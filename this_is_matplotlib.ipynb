```python
import matplotlib.pyplot as plt
marks = [3,2,6,7,9,0,10,11,7,5,4,14]
#range -- 0-5,5-10,10-15-------bins
plt.hist(marks,bins=[0,5,10,15,20],color= "pink")
plt.xlabel ("marks distribution ")
plt.ylabel("frequency")
plt.title("marks distribution");
plt.show()
```

```python
#pie chart
marks =[20,30,10,30,7,3]
subject= ["cse","math","eng","EEE", "IT", "art"]
```

```python
plt.pie(marks,labels= subject,autopct="%0.1f%%", colors=["pink", "aqua", "azure", "lightblue", "lightyellow","red"], explode=[0.1, 0.2, 0.1 , 0.2 , 0.3 , 0.5])
plt.show()
```
